# **Problem Statement**

---



**Multi-line sequence fasta to single-line fasta file**

Fasta file format is shown below. Many times, it is annoying to find sequences of interest because of fix-width format, i.e., if 60 character fix-width file is there then only except the header (starting with >) of the sequence each line will have 60 characters. So, if a sequence is of width 600, then it would be written in 10 lines. Your goal is convert this format to a format with header in a single line followed by its sequence in a single line.

# **Approach**

---


FASTA files often store DNA or protein sequences in multiple fixed-width lines, which can make pattern searching across line breaks challenging. To address this:
- Header Detection:
  - The code reads the FASTA file line by line.
  -	If a line starts with >, it is recognized as a header and written directly to the output.
- Sequence Handling:
	-	For all sequence lines (those not starting with >), the code removes line breaks and concatenates them into a single continuous string.
	-	This process is repeated for each sequence in the file.
-	Output Structure:
The result is a streamlined FASTA file where:
	-	Each header line (>...) is followed by a single, uninterrupted sequence line.
	-	This structure improves readability and allows for efficient pattern matching and other analyses without worrying about line breaks within sequences.


```python
# Convert multi-line FASTA to single-line per sequence format

input_path = "multiline_input.fasta"
output_path = "q3_output.fasta"

with open(input_path, 'r') as infile, open(output_path, 'w') as outfile:
    first_header = True
    for raw_line in infile:
        line = raw_line.strip()

        if line.startswith('>'):
            if not first_header:
                outfile.write('\n')  # Finish previous sequence
                print()             # Print newline after previous sequence
            outfile.write(line + '\n')
            print(line)
            first_header = False
        else:
            outfile.write(line)
            print(line, end='')  # Avoid extra newline while printing sequence

    outfile.write('\n')  # Final newline
    print()              # Final newline for clean output

print(f"\nReformatted file saved as: '{output_path}'")
```
