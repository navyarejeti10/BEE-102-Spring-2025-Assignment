**Building Markov transition matrix**

You have learned important differences between simple dinucleotide frequency model and Markov models. Your goal is to build one from the input sequence. In the Jupyter notebook, you should print the transition matrix.

# **Approach**

---
To compute the Markov transition matrix from a DNA sequence, we follow these steps:
#### **1. Count Dinucleotide Transitions**
- We iterate through the DNA sequence one nucleotide at a time, examining each pair of consecutive bases (e.g., 'AG', 'CT', 'TG').
For each pair:
  -	The first nucleotide represents the current state.
  -	The second nucleotide represents the next state.
#### **2. Store Counts in a Nested Dictionary**
- We maintain a nested dictionary structure where:
  -	Rows correspond to the current nucleotide.
  -	Columns represent the next nucleotide.
  - Each cell contains the count of transitions from one base to another.
####	**3.	Normalize to Get Probabilities**
- After collecting all counts:
  -	We convert them into probabilities by dividing each count by the total number of transitions originating from the same current nucleotide.
  -	This normalization ensures each row in the matrix sums to 1, forming a valid Markov transition matrix.

Finally, we represent the transition matrix as a Pandas DataFrame for easier visualization and interpretation.


```python
import pandas as pd

# Given DNA sequence for analysis
dna_sequence = "AGTCCAGTCTAGGTCGATCGTAGGCTAGCTAGCTGACTGAGCTAGCATCGGATCGTA"

# Valid nucleotide states
nucleotides = ['A', 'C', 'G', 'T']
```

```python
# Initialize a nested dictionary to hold dinucleotide counts
transition_counts = {
    base: {next_base: 0 for next_base in nucleotides}
    for base in nucleotides
}

# Iterate over the sequence and count base-to-base transitions
for idx in range(len(dna_sequence) - 1):
    current = dna_sequence[idx]
    next_base = dna_sequence[idx + 1]

    if current in nucleotides and next_base in nucleotides:
        transition_counts[current][next_base] += 1
    else:
        print(f"Skipping invalid base: {current} or {next_base}")
```

```python
# Convert raw counts to transition probabilities
markov_probabilities = {
    base: {} for base in nucleotides
}

for base in nucleotides:
    total = sum(transition_counts[base].values())
    for next_base in nucleotides:
        if total > 0:
            prob = transition_counts[base][next_base] / total
        else:
            prob = 0.0
        markov_probabilities[base][next_base] = prob
```

```python
# Convert the probability dictionary to a readable DataFrame
transition_df = pd.DataFrame(markov_probabilities)

print("Markov Transition Matrix:")
print("Rows: Current base, Columns: Next base")
print(transition_df.round(4))
```
